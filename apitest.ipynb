```python
import os
from dotenv import load_dotenv

load_dotenv()


BAZZAAR_API_KEY = os.getenv("BAZARRLINK")
```

```python
from openai import OpenAI

client = OpenAI(base_url="https://api.bazaar.ai/v1", api_key=BAZZAAR_API_KEY)

#sample query
query = "What is the syllabus covered in Modern LLM finetuning?"

response = client.chat.completions.create(
    model="deepseek/deepseek-v4-flash-0731free",
    messages=[
        {"role": "user", "content": query}
    ]
)

print(response)
```

```python
from openai import OpenAI

client = OpenAI(
    api_key=BAZZAAR_API_KEY,
    base_url="https://api.bazaarlink.ai/v1",
    timeout=60,
)

models = client.models.list()

for model in models.data:
    print(model.id)
```
